```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns

df= pd.read_csv("Customer Churn.csv")
df.head()
```

```python
df.info()
```

```python
df['TotalCharges'] =df['TotalCharges'].replace(" ","0")
df['TotalCharges'] =df['TotalCharges'].astype("float")
```

```python
def conv(value):
    if value == 1:
        return "Yes"
    else:
        return "No"
df["SeniorCitizen"] =df["SeniorCitizen"].apply(conv)       
```

```python
ax=sns.countplot(data=df, x= "Churn")
plt.title("Count of Custumer by Churn", fontsize = 15)
plt.grid()
for bars in ax.containers:
    ax.bar_label(bars)
    
```

```python
gb= df.groupby("Churn").agg({"Churn":"count"})

plt.figure(figsize =(10,5))
plt.pie(gb["Churn"], labels = gb.index, autopct = "%1.2f%%")
plt.title("Percentage of Churned Customeres", fontsize = 15)
plt.show()
```

```python
ax=sns.countplot(data= df, x= 'gender', hue= 'Churn')
plt.title("Churn By an Gender", fontsize = 15)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
ax=sns.countplot(data= df, x= 'SeniorCitizen', hue= 'Churn')
plt.title("Churn By an Gender", fontsize = 15)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Create the countplot
ax = sns.countplot(data=df, x="SeniorCitizen", hue="Churn")
plt.title("Churn by Senior Citizen Status", fontsize=15)

# Calculate percentages and add labels
for bars in ax.containers:
    # Calculate the percentage for each bar based on the total dataframe size
    labels = [f"{(bar.get_height() / len(df)) * 100:.1f}%" for bar in bars]
    ax.bar_label(bars, labels=labels)

plt.show()
```

```python
ax=sns.histplot(data= df,x= "tenure",bins= 10, hue ='Churn')

plt.title("Churn By an month", fontsize = 15)
plt.grid()
plt.figure(figsize =(9,4))
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
import math
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Aapke saare columns ki list
features = [
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
]
target = "Churn"

# 2. 3x3 ki subplots grid setup karein
n_cols = 3
n_rows = 3
fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 5 * n_rows))
axes = axes.flatten()  # Grid ko loop chalane ke liye flatten kiya

# 3. Loop chala kar har column ka countplot aur percentage label lagayein
for i, feature in enumerate(features):
    ax = sns.countplot(data=df, x=feature, hue=target, ax=axes[i])
    ax.set_title(f"Churn by {feature}", fontsize=14)
    ax.set_xlabel(feature, fontsize=12)
    ax.set_ylabel("Count", fontsize=12)

    # Label ko percentage mein badalne ke liye loop
    for bars in ax.containers:
        # None values ya NaN se error na aaye isliye checks lagaye hain
        labels = []
        for bar in bars:
            height = bar.get_height()
            if height > 0:
                percentage = (height / len(df)) * 100
                labels.append(f"{percentage:.1f}%")
            else:
                labels.append("")  # Agar count 0 hai toh khali chhod dein
        ax.bar_label(bars, labels=labels)

# 4. Plots overlap na karein isliye layout adjust kiya
plt.tight_layout()
plt.show()
```
